```python
import sympy as sp
```

```python
x = sp.Symbol('x')
y = sp.Function('y')(x)
```

```python
print("a)", sp.simplify(x*(-x + sp.cos(x)) + x))
```

```python
ode = sp.Eq(y.diff(x), -2*y/x)
sol = sp.dsolve(ode, y)
C1 = sp.Symbol('C1')
y_sol = sol.rhs
C1_val = sp.solve(y_sol.subs(x, 2) - 2, C1)[0]
print("b)", sp.Eq(y, y_sol.subs(C1, C1_val)))
```
