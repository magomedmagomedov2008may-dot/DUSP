```python
from sympy import *
```

```python
t = symbols("t", positive=True)
k = symbols("k", positive=True)
m0 = symbols("m0", positive=True)
m = Function("m")
```

```python
ode = Eq(diff(m(t), t), -k * m(t))
print("Уравнение:", ode)
```

```python
sol = dsolve(ode, ics={m(0): m0})
print("Решение:", sol)
```

```python
eq_half = Eq(m0 / 2, sol.rhs.subs(t, 1590))
real_k = solve(eq_half, k)[0]
print("k =", real_k)
```

```python
sol_final = sol.subs(k, real_k)
print("m(t) =", sol_final)
```

```python
m_200 = sol_final.rhs.subs(t, 200)
print("m(200) =", m_200)
```

```python
percent_decayed = (m0 - m_200) / m0 * 100
print("Процент распада =", percent_decayed.evalf())
```
