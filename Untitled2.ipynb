```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = sp.Symbol('x')
y = sp.Function('y')(x)
y_expr = x**2
```

```python
result = sp.integrate(y_expr, x)
print("Интеграл:", result)
```

```python
f_original = sp.lambdify(x, y_expr, 'numpy')
f_integral = sp.lambdify(x, result, 'numpy')
```

```python
x_vals = np.linspace(-3, 3, 400)
```

```python
plt.figure(figsize=(10, 6))
```

```python
plt.plot(x_vals, f_original(x_vals), label='y = x²', color='blue', linewidth=2)
```

```python
plt.plot(x_vals, f_integral(x_vals), label='∫y dx = x³/3', color='red', linewidth=2, linestyle='--')
```

```python
plt.axhline(0, color='black', linewidth=0.5) # Ось X
plt.axvline(0, color='black', linewidth=0.5) # Ось Y
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=12)
plt.title('Исходная функция и её интеграл', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.ylim(-10, 10)
```

```python
plt.show()
```
